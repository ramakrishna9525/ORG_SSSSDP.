```python
n = int(input('Enter number of lines you need : '))
for i in range(1,n+1):
    print(" " * (n-i), end = '')
    for j in range(65, 65+i):
        print(chr(j), end = ' ')
    print()
```

```python
n = int(input('Enter number of lines you need : '))
for i in range(1, n+1):
    print(' ' * (n-i), end='')
    for j in range(1, 2*i):
        if n == i:
            print('*', end='')
        elif j == 1 or j == 2 * i - 1:
            print('*', end='')
        else:
            print(' ', end='')
    print()
```
